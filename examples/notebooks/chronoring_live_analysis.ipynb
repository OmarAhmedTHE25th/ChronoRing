# ChronoRing Live Columnar Stream Analysis (Jupyter + Polars + Arrow)

This notebook demonstrates reading ChronoRing's `.crng` columnar telemetry stream incrementally via `chronoring.live_stream`, converting zero-copy column slices to **Polars** or **PyArrow**, and hot-reloading trigger configuration via the Dashboard Control Plane API.

```python
import sys
sys.path.insert(0, '../../python')

from chronoring.live_stream import LiveCrngStream, read_crng_columns, to_polars_dataframe, to_pyarrow_table
```

```python
# 1. Incremental live streaming of committed 1024-event `.crng` pages (uncompressed or LZ4-compressed)
stream = LiveCrngStream('../../demo.crng')
for page in stream.iter_available():
    print(f'Received {len(page["sequence_id"])} events, mean energy={sum(page["energy_mev"])/len(page["energy_mev"]):.2f} MeV')
```
